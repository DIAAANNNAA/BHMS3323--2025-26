```
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

## Step 1: Download Tesla (TSLA) Stock Data

We will use the `yfinance` library to download Tesla's stock data for the year 2024.


```
# Download Tesla stock data for 2024
data = yf.download("TSLA", start="2024-01-01", end="2024-12-31")

# Display the first few rows
data.head()
```

## Step 2: Calculate Moving Averages

We will calculate:
- 20-day moving average (MA20)
- 50-day moving average (MA50)


```
# Calculate moving averages
data['MA20'] = data['Close'].rolling(window=20).mean()
data['MA50'] = data['Close'].rolling(window=50).mean()

# Display the data with moving averages
data[['Close', 'MA20', 'MA50']].head(30)
```

## Step 3: Plot Stock Price and Moving Averages

We will visualize Tesla's closing price along with MA20 and MA50.


```
# Plot the stock price and moving averages
plt.figure(figsize=(12,6))
plt.plot(data['Close'], label='Close Price', color='blue')
plt.plot(data['MA20'], label='MA20', color='orange')
plt.plot(data['MA50'], label='MA50', color='green')
plt.title('Tesla (TSLA) Stock Price with MA20 and MA50')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()
```

## Step 4: Implement a Simple Strategy

Our strategy:
- **Buy Signal**: When MA20 crosses above MA50 (i.e., MA20 > MA50 and previous day MA20 <= MA50)


```
# Implement the buy signal strategy
data['Signal'] = (data['MA20'] > data['MA50']) & (data['MA20'].shift(1) <= data['MA50'].shift(1))

# Extract buy signal dates
buy_signal_dates = data.index[data['Signal']].strftime('%Y-%m-%d').tolist()
print("Buy Signal Dates:", buy_signal_dates)
```
